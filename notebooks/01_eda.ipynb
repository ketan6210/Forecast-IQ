# ForecastIQ: data exploration
Synthetic retail demo only. All values below are calculated from the bundled CSV.

```python
from pathlib import Path
import sys
import pandas as pd
root = Path.cwd() if (Path.cwd() / "backend").exists() else Path.cwd().parent
sys.path.insert(0, str(root))
from backend.app.data.loader import read_csv
from backend.app.data.validator import DataValidator
frame, report = DataValidator().validate(read_csv((root / "data/demo/retail_sales.csv").read_bytes()))
report.model_dump()
```

```python
frame.head()
```

```python
frame.groupby("store_id").agg(units=("units_sold", "sum"), products=("product_id", "nunique"))
```

```python
daily = frame.groupby("date").units_sold.sum()
daily.groupby(daily.index.dayofweek).mean().rename("average_daily_units")
```

```python
daily.groupby(daily.index.month).mean().rename("monthly_average_daily_units")
```
